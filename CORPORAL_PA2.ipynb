# REPRODUCIBLE NORMALIZATION PROBLEM

```python
import numpy as np

np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))

mean = X.mean()
standarddeviation = X.std()
X_normalized = (X - mean)/standarddeviation

print("X = \n", X)

print("\nNormalized = \n", X_normalized)

print("\nMean = \n", mean)

print("\nStandard Deviation = \n", standarddeviation)

np.save('X_normalized.npy', X_normalized)
```

# CUBES DIVISIBLE BY 4 PROBLEM

```python
import numpy as np

C = np.arange(1, 101)**3
C = C.reshape(10, 10)

div_by_4 = C[C % 4 == 0]

print("Shape of C = \n", C.shape)
print("\nC = \n", C)
print("\nArray of div_by_4 = \n", div_by_4)
print("\nNumber of Selected Elements\n", div_by_4.size)

np.save('div_by_4.npy', div_by_4)
```

# ABOVE-SQUARES PROBLEM

```python
import numpy as np

S = np.arange(1, 37)**2
S = S.reshape(6, 6)
S_mean = S.mean()
above_mean = S[S > S_mean]

print("S = \n", S)
print("\nThe mean of S = \n", S_mean)
print("\nElements above the mean of S = \n", above_mean)
print("\nNumber of elements above the mean of S = \n", above_mean.size)

np.save('above_mean.npy', above_mean)
```
